```
# Part 1: Matplotlib
import matplotlib.pyplot as plt
import pandas as pd

data = {'Month': ['Jan','Feb','Mar','Apr','May'],
        'Sales': [200, 250, 300, 400, 350]}
df = pd.DataFrame(data)

plt.figure(figsize=(8,5))
plt.plot(df['Month'], df['Sales'], marker='o', color='blue')
plt.title('Monthly Sales - Line Plot')
plt.xlabel('Month')
plt.ylabel('Sales')
plt.grid(True)
plt.show()

plt.figure(figsize=(8,5))
plt.bar(df['Month'], df['Sales'], color='green')
plt.title('Monthly Sales - Bar Chart')
plt.xlabel('Month')
plt.ylabel('Sales')
plt.show()

# Part 2: Seaborn
import seaborn as sns
tips = sns.load_dataset("tips")

sns.scatterplot(data=tips, x="total_bill", y="tip", hue="sex")
plt.title("Scatter Plot: Total Bill vs Tip")
plt.show()

sns.boxplot(data=tips, x="day", y="total_bill")
plt.title("Box Plot: Total Bill by Day")
plt.show()

# Part 3: Folium
!pip install folium
import folium
map_ahm = folium.Map(location=[23.0225, 72.5714], zoom_start=11)
folium.Marker(location=[23.0225, 72.5714], popup="Ahmedabad, Gujarat", icon=folium.Icon(color='red')).add_to(map_ahm)
map_ahm
```
